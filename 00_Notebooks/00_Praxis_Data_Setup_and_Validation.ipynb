**Notebook 00: Data Setup and Validation**


Step 1: Set Up Git Environment

GitHub was used as the Python environment for the analysis. Google Drive was connected so the project folders and dataset files could be accessed throughout the coding process.

```python
# Step 1: Set up Git Environment

# Load the libraries
import os
import pandas as pd

# Set the main project folder
main_folder = os.getcwd()

# Check the project folder
if not os.path.isdir(main_folder + "/01_Simulated_RFID_Data"):
    main_folder = os.path.dirname(main_folder)

print("GitHub Codespaces environment is ready.")
```

Step 2: Connect the Praxis Data

The main project folder and the five dataset folders were identified so the files could be accessed directly from Google Drive. The simulated RFID data is the primary aviation dataset, while the other datasets provide supporting cybersecurity data.

```python
# Step 2: Connect the Praxis data

# Set the dataset folders

rfid_folder = main_folder + "/01_Simulated_RFID_Data"
ciciot_folder = main_folder + "/02_CICIoT2023"
nbaiot_folder = main_folder + "/03_N_BaIoT"
iot23_folder = main_folder + "/04_IoT_23"
cicids_folder = main_folder + "/05_CICIDS2017"

print("The Praxis data folders are connected.")
```

Step 3: Confirm the Dataset Files

The dataset folders were checked to confirm that the files were available before the analysis began. The number of total files and CSV files was reviewed for each dataset.

```python
# Step 3: Confirm the dataset files

dataset_names = [
    "Simulated RFID Data",
    "CICIoT2023",
    "N BaIoT",
    "IoT 23",
    "CICIDS2017"
]

dataset_folders = [
    rfid_folder,
    ciciot_folder,
    nbaiot_folder,
    iot23_folder,
    cicids_folder
]

# Check each dataset folder

for number in range(len(dataset_folders)):

    dataset_name = dataset_names[number]
    dataset_folder = dataset_folders[number]

    file_count = 0
    csv_count = 0

    for current_folder, folders, files in os.walk(dataset_folder):

        for file_name in files:

            file_count = file_count + 1

            if file_name.endswith(".csv"):
                csv_count = csv_count + 1

    print(dataset_name)
    print("Total files:", file_count)
    print("CSV files:", csv_count)
    print()
```

Step 4: Create the Dataset Inventory

A basic inventory was created to document the files available for each dataset. The inventory records the dataset name, file name, file location, file type, and file size before any data preparation or analysis is completed.

```python
# Step 4: Create the dataset inventory

inventory_list = []

# Check each dataset

for number in range(len(dataset_folders)):

    dataset_name = dataset_names[number]
    dataset_folder = dataset_folders[number]

    for current_folder, folders, files in os.walk(dataset_folder):

        for file_name in files:

            file_path = os.path.join(current_folder, file_name)
            file_type = os.path.splitext(file_name)[1]
            file_size = os.path.getsize(file_path)
            file_size_mb = file_size / (1024 * 1024)

            inventory_list.append([
                dataset_name,
                file_name,
                file_path,
                file_type,
                round(file_size_mb, 2)
            ])

# Create the inventory table

inventory = pd.DataFrame(
    inventory_list,
    columns=[
        "Dataset Name",
        "Original Filename",
        "File Location",
        "File Type",
        "File Size MB"
    ]
)

print("Total files in the inventory:", len(inventory))

inventory.head()
```

Step 5: Validate the Primary RFID Dataset

The simulated RFID baggage tracking file was reviewed to confirm that it loaded correctly and contained the expected data structure. The number of records, columns, missing values, and duplicate records were checked before the data was prepared for analysis.

```python
# Step 5: Validate the primary RFID dataset

# Load the RFID file

rfid_file = rfid_folder + "/01_Simulated_RFID_Data.csv"

rfid_data = pd.read_csv(rfid_file)

# Check the dataset

print("Rows:", rfid_data.shape[0])
print("Columns:", rfid_data.shape[1])
print("Missing values:", rfid_data.isnull().sum().sum())
print("Duplicate records:", rfid_data.duplicated().sum())

# Show the column names

print("\nColumn names:")
print(rfid_data.columns.tolist())

# Show the first five records

rfid_data.head()
```

Step 6: Validate the Supporting Datasets

A sample CSV file from each supporting cybersecurity dataset was opened to confirm that the files could be read in Google Colab. The supporting datasets remained separate from the simulated RFID baggage data. The dataset inventory was then saved for use during the remaining analysis.

```python
# Step 6: Validate the supporting datasets

supporting_datasets = [
    "CICIoT2023",
    "N BaIoT",
    "IoT 23",
    "CICIDS2017"
]

# Check one CSV file from each dataset

for dataset_name in supporting_datasets:

    dataset_files = inventory[
        (inventory["Dataset Name"] == dataset_name) &
        (inventory["File Type"] == ".csv")
    ]

    # Use a benign traffic file for N BaIoT when available

    if dataset_name == "N BaIoT":

        benign_files = dataset_files[
            dataset_files["Original Filename"] == "benign_traffic.csv"
        ]

        if len(benign_files) > 0:
            file_path = benign_files.iloc[0]["File Location"]
        else:
            file_path = dataset_files.iloc[0]["File Location"]

    else:
        file_path = dataset_files.iloc[0]["File Location"]

    # Open a small sample

    sample_data = pd.read_csv(file_path, nrows=5)

    print(dataset_name)
    print("File:", os.path.basename(file_path))
    print("Columns:", len(sample_data.columns))
    print("File opened successfully.")
    print()

# Save the dataset inventory

inventory_file = main_folder + "/09_Tables/dataset_inventory.csv"

inventory.to_csv(inventory_file, index=False)

print("Dataset inventory saved.")
```
