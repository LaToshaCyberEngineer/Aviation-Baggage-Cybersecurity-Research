
# H2 N BaIoT Machine Learning Analysis

**H2: A Random Forest model provides the highest spoofing detection accuracy in IoT baggage tracking data compared to other ML models.**

**RQ2: Which ML models are most effective at detecting spoofing in Internet of Things enabled baggage tracking data?**

This notebook extends the H2 Machine Learning analysis by comparing Logistic Regression, Support Vector Machine, and Random Forest using the N BaIoT cybersecurity dataset. The analysis examines normal IoT device activity and malicious botnet traffic separately from the simulated RFID baggage tracking records.

**BLUF:** The analysis compares the three Machine Learning models to determine which provides the strongest detection performance using N BaIoT network traffic. The findings supplement the original H2 experiment but do not directly measure RFID baggage reader spoofing.


```python

# Step 1: Connect and review the N BaIoT data

# Load the libraries
import os
import pandas as pd


# Set the main project folder
main_folder = os.getcwd()

if not os.path.isdir(main_folder + "/03_N_BaIoT"):
    main_folder = os.path.dirname(main_folder)


# Set the N BaIoT data folder
data_folder = main_folder + "/03_N_BaIoT/data/extracted"


# Find the CSV files
csv_files = []

for folder, subfolders, filenames in os.walk(data_folder):
    for filename in filenames:
        if filename.endswith(".csv"):
            csv_files.append(os.path.join(folder, filename))


# Sort the filenames
csv_files.sort()

print("N BaIoT dataset connected.")

print("\nTotal CSV files:", len(csv_files))


# Identify normal traffic files
benign_files = [
    file for file in csv_files
    if "benign" in file.lower()
]


# Identify botnet attack files
attack_files = [
    file for file in csv_files
    if "mirai" in file.lower() or "gafgyt" in file.lower()
]


# Display the file counts
print("Normal traffic files:", len(benign_files))
print("Botnet attack files:", len(attack_files))


# Review a sample normal traffic file
if benign_files:

    benign_sample = pd.read_csv(benign_files[0], nrows=5)

    print("\nNormal traffic sample file:")
    print(os.path.relpath(benign_files[0], main_folder))

    print("\nAvailable fields:")
    print(benign_sample.columns.tolist())

    print("\nSample normal traffic records:")
    print(benign_sample.head())


# Review a sample attack traffic file
if attack_files:

    attack_sample = pd.read_csv(attack_files[0], nrows=5)

    print("\nAttack traffic sample file:")
    print(os.path.relpath(attack_files[0], main_folder))

    print("\nAvailable fields:")
    print(attack_sample.columns.tolist())

    print("\nSample attack traffic records:")
    print(attack_sample.head())
```
