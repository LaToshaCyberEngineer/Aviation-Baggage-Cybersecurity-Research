H3: Cybersecurity analytics reduces the operational impact of simulated denial of service conditions in RFID baggage tracking environments.

This supplemental analysis uses the N BaIoT dataset to evaluate machine learning detection of malicious IoT network activity. The experiment compares Logistic Regression, Support Vector Machine, and Random Forest using recorded normal traffic and Mirai and Gafgyt botnet attacks.

The analysis uses device grouped validation to determine whether the models can identify malicious traffic from IoT devices excluded from training. A majority class baseline is included to evaluate whether the trained models provide improved detection performance.

The experiment supplements H3 by examining network attack detection. It does not directly measure baggage scan loss, operational delays, or recovery following a denial of service attack.

Step 3: Validate Data Structure and Features

The available files are organized by IoT device and traffic category. File labels and numerical features are examined before model development.

The validation checks whether benign and malicious traffic are available for each device and whether the CSV files share a consistent feature structure.

File names are used to identify attack categories. Device identifiers and file names are retained as experimental metadata rather than model predictors.

```python
# Step 3: Validate the N BaIoT data structure

import os
import re
import pandas as pd

from pathlib import Path

# Convert the existing CSV paths into Path objects.
csv_paths = [Path(file) for file in csv_files]

if not csv_paths:
    raise ValueError("No N BaIoT CSV files were found.")

# Find the shared parent folder.
common_folder = Path(
    os.path.commonpath([str(file.parent) for file in csv_paths])
)

# Store information about each file.
file_records = []

for file in csv_paths:

    # Identify the relative location of the file.
    relative = file.relative_to(common_folder)
    location = relative.as_posix().lower()

    # Identify the traffic class and botnet family.
    if "benign" in location:
        traffic_class = "benign"
        family = "benign"

    elif "mirai" in location:
        traffic_class = "attack"
        family = "mirai"

    elif "gafgyt" in location or "bashlite" in location:
        traffic_class = "attack"
        family = "gafgyt"

    else:
        traffic_class = "unclassified"
        family = "unknown"

    # Identify the device from the folder structure.
    if len(relative.parts) > 1:
        device = relative.parts[0]

    else:
        # Support filenames containing device and attack labels.
        device = re.split(
            r"[._-](?=benign|mirai|gafgyt|bashlite)",
            file.stem,
            maxsplit=1,
            flags=re.IGNORECASE
        )[0]

    file_records.append({
        "path": str(file),
        "relative_path": relative.as_posix(),
        "device": device,
        "traffic_class": traffic_class,
        "family": family
    })

# Create the file inventory.
file_index = pd.DataFrame(file_records)

# Display files without recognized labels.
unclassified = file_index[
    file_index["traffic_class"] == "unclassified"
]

print("Unclassified files:", len(unclassified))

if not unclassified.empty:
    print(unclassified[["relative_path"]].to_string(index=False))

# Use only files with recognized traffic labels.
file_index = file_index[
    file_index["traffic_class"].isin(["benign", "attack"])
].copy()

# Count normal and attack files for each device.
device_inventory = pd.crosstab(
    file_index["device"],
    file_index["traffic_class"]
)

print("\nDevice and traffic inventory:")
print(device_inventory)

# Verify that each device has both traffic categories.
eligible_devices = device_inventory[
    (device_inventory.get("benign", 0) > 0) &
    (device_inventory.get("attack", 0) > 0)
].index.tolist()

print("\nDevices with benign and attack files:", len(eligible_devices))

# Read column headers without loading entire files.
schemas = {}

for file in file_index["path"]:

    columns = tuple(pd.read_csv(file, nrows=0).columns)

    schemas.setdefault(columns, []).append(file)

print("Distinct CSV feature schemas:", len(schemas))

# Stop if the CSV files have inconsistent columns.
if len(schemas) != 1:
    raise ValueError(
        "The CSV files have different column structures. "
        "Review the schemas before proceeding."
    )

# Preserve the verified feature names.
feature_cols = list(next(iter(schemas)))

print("Number of available features:", len(feature_cols))
print("First five features:", feature_cols[:5])

# Check a small sample for numerical values.
sample_check = pd.read_csv(
    file_index.iloc[0]["path"],
    nrows=10
)

non_numeric = [
    column for column in feature_cols
    if not pd.api.types.is_numeric_dtype(sample_check[column])
]

if non_numeric:
    raise ValueError(
        f"Non-numeric features require review: {non_numeric}"
    )

# Require enough eligible devices for grouped evaluation.
if len(eligible_devices) < 3:
    raise ValueError(
        "At least three devices with both classes are required."
    )

print("\nInitial file and feature validation completed.")
```

Step 4: Prepare the Experimental Sample

The N BaIoT dataset contains substantially more records than are necessary for an initial supplemental experiment.

A reproducible sample is collected across each eligible CSV file using a fixed random seed. The procedure reads the files in smaller portions to limit memory use.

Benign and malicious observations are then balanced within each device. This allows model detection performance to be evaluated without allowing the larger attack class to dominate the sample.

The resulting class proportions are part of the experimental design and do not represent the natural frequency of attacks in operational networks.

```python
# Step 4: Prepare a reproducible N BaIoT sample

import numpy as np
import pandas as pd

# Set the experimental random seed.
RANDOM_SEED = 42

# Limit the number of records collected from each CSV.
SAMPLE_PER_FILE = 400

# Process files in smaller portions.
CHUNK_SIZE = 20000

# Retain only eligible devices.
selected_files = file_index[
    file_index["device"].isin(eligible_devices)
].sort_values("relative_path").reset_index(drop=True)


def sample_csv_file(file_path, sample_size, seed):
    # Create a reproducible random generator.
    rng = np.random.default_rng(seed)

    # Store the running sample.
    reservoir = None
    rows_seen = 0

    # Read the CSV in memory-limited portions.
    reader = pd.read_csv(
        file_path,
        chunksize=CHUNK_SIZE,
        dtype=np.float32
    )

    for chunk in reader:

        # Count records in the current portion.
        new_rows = len(chunk)

        # Determine the required sample size.
        required = min(
            sample_size,
            rows_seen + new_rows
        )

        # Randomly determine how many records to retain
        # from the new portion.
        take_new = int(
            rng.hypergeometric(
                new_rows,
                rows_seen,
                required
            )
        )

        take_old = required - take_new

        # Select records from the new portion.
        new_sample = chunk.iloc[
            rng.choice(
                new_rows,
                size=take_new,
                replace=False
            )
        ]

        # Retain the selected portion of the old sample.
        if reservoir is not None and take_old > 0:
            old_sample = reservoir.iloc[
                rng.choice(
                    len(reservoir),
                    size=take_old,
                    replace=False
                )
            ]

            reservoir = pd.concat(
                [old_sample, new_sample],
                ignore_index=True
            )

        else:
            reservoir = new_sample.reset_index(drop=True)

        # Update the number of processed records.
        rows_seen += new_rows

    return reservoir


# Collect samples without modifying the original CSV files.
sample_parts = []

for file_number, row in selected_files.iterrows():

    print(
        f"Sampling file {file_number + 1} "
        f"of {len(selected_files)}: "
        f"{row['relative_path']}"
    )

    file_sample = sample_csv_file(
        row["path"],
        SAMPLE_PER_FILE,
        RANDOM_SEED + file_number
    )

    if file_sample is None or file_sample.empty:
        continue

    # Add labels and metadata after sampling.
    file_sample["device"] = row["device"]
    file_sample["family"] = row["family"]
    file_sample["traffic_class"] = row["traffic_class"]
    file_sample["source_file"] = row["relative_path"]

    # Use 0 for normal and 1 for malicious traffic.
    file_sample["label"] = (
        0 if row["traffic_class"] == "benign" else 1
    )

    sample_parts.append(file_sample)

if not sample_parts:
    raise ValueError("No valid sample records were collected.")

sampled_data = pd.concat(
    sample_parts,
    ignore_index=True
)

# Balance normal and attack observations within each device.
balanced_parts = []

for device, group in sampled_data.groupby("device"):

    normal = group[group["label"] == 0]
    attack = group[group["label"] == 1]

    # Use the smaller category count for both classes.
    count = min(len(normal), len(attack))

    if count == 0:
        continue

    balanced_parts.append(
        normal.sample(n=count, random_state=RANDOM_SEED)
    )

    balanced_parts.append(
        attack.sample(n=count, random_state=RANDOM_SEED)
    )

# Create the final experimental sample.
sample_df = pd.concat(
    balanced_parts,
    ignore_index=True
)

# Replace infinite values with missing values.
# Missing values will be handled inside training pipelines.
sample_df[feature_cols] = sample_df[
    feature_cols
].replace([np.inf, -np.inf], np.nan)

print("\nPrepared sample records:", len(sample_df))

print("\nTraffic class distribution:")
print(sample_df["traffic_class"].value_counts())

print("\nRecords by device:")
print(pd.crosstab(sample_df["device"], sample_df["label"]))
```

Step 5: Establish Device Held Out Validation

The experimental sample is divided using device identifiers rather than randomly distributing individual records between training and testing.

Three-fold grouped validation is used so that every device appears in a test partition without appearing in the corresponding training partition.

This design evaluates whether a model can identify malicious traffic from previously unseen devices. It reduces device-specific data leakage but does not guarantee independence between all recorded attack conditions.

```python
# Step 5: Establish device held out validation

from sklearn.model_selection import GroupKFold

# Select only the verified numerical predictors.
X = sample_df[feature_cols].copy()

# Select the binary attack classification target.
y = sample_df["label"].copy()

# Use device identity only for grouping.
groups = sample_df["device"]

# Create three nonoverlapping device groups.
grouped_cv = GroupKFold(n_splits=3)

# Preserve the folds for every model comparison.
folds = list(
    grouped_cv.split(X, y, groups=groups)
)

# Review each training and test division.
for fold_number, (train_idx, test_idx) in enumerate(
    folds,
    start=1
):

    train_devices = set(groups.iloc[train_idx])
    test_devices = set(groups.iloc[test_idx])

    # Confirm that devices do not overlap.
    overlap = train_devices.intersection(test_devices)

    if overlap:
        raise ValueError("Device leakage was detected.")

    # Compare botnet families in training and testing.
    train_families = set(
        sample_df.iloc[train_idx]["family"]
    ) - {"benign"}

    test_families = set(
        sample_df.iloc[test_idx]["family"]
    ) - {"benign"}

    print(f"\nFold {fold_number}")

    print("Training devices:", sorted(train_devices))
    print("Testing devices:", sorted(test_devices))

    print("Training records:", len(train_idx))
    print("Testing records:", len(test_idx))

    print("Training attack families:", sorted(train_families))
    print("Testing attack families:", sorted(test_families))

    # Stop if a test family is absent from training.
    if not test_families.issubset(train_families):
        raise ValueError(
            "An attack family is absent from training. "
            "Review the experimental design."
        )

print("\nDevice separation checks completed.")
```


Step 6: Train and Compare Machine Learning Models

Logistic Regression, Support Vector Machine, and Random Forest are trained using the prepared N BaIoT data. A majority class baseline is included for comparison.

The models use the same device grouped validation folds to evaluate their ability to detect malicious network activity on unfamiliar IoT devices. Predictions are retained for the performance evaluation in Step 7.

```python
# Step 6: Train and Compare Machine Learning Models

# Import the tools needed to prepare and evaluate the models.
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import GroupKFold
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Import the machine learning models.
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier


# Check that Step 4 prepared the required data.
if "sample_df" not in globals():
    raise RuntimeError(
        "Prepared data is missing. Run Step 4 before Step 6."
    )

if "feature_cols" not in globals():
    raise RuntimeError(
        "Feature names are missing. Run Step 3 before Step 6."
    )

# Set the same random seed used in Step 4.
RANDOM_SEED = 42


# Select the numerical features used for model training.
X = sample_df[feature_cols].copy()

# Select the known traffic labels.
# 0 represents benign traffic.
# 1 represents malicious botnet traffic.
y = sample_df["label"].copy()

# Identify devices for grouped validation.
# Device names are not used as prediction features.
groups = sample_df["device"].copy()


# Confirm that enough devices are available.
if groups.nunique() < 3:
    raise ValueError(
        "At least three devices are required for validation."
    )


# Create three device grouped validation folds.
# This reproduces the splitting method from Step 5.
grouped_cv = GroupKFold(n_splits=3)

folds = list(
    grouped_cv.split(X, y, groups=groups)
)


# Verify the device separation before training.
for fold_number, (train_idx, test_idx) in enumerate(
    folds,
    start=1
):

    # Identify the devices assigned to each group.
    train_devices = set(groups.iloc[train_idx])
    test_devices = set(groups.iloc[test_idx])

    # Prevent the same device from appearing in both groups.
    if train_devices.intersection(test_devices):
        raise ValueError(
            f"Device overlap detected in Fold {fold_number}."
        )

    # Verify that both classes exist in each partition.
    if y.iloc[train_idx].nunique() != 2:
        raise ValueError(
            f"Training Fold {fold_number} is missing a class."
        )

    if y.iloc[test_idx].nunique() != 2:
        raise ValueError(
            f"Testing Fold {fold_number} is missing a class."
        )

    # Identify attack families in each partition.
    train_families = set(
        sample_df.iloc[train_idx]["family"]
    ) - {"benign"}

    test_families = set(
        sample_df.iloc[test_idx]["family"]
    ) - {"benign"}

    # Verify that testing does not introduce an untrained family.
    if not test_families.issubset(train_families):
        raise ValueError(
            f"Fold {fold_number} has an unseen attack family."
        )

print("Device grouped validation confirmed.")
print("Number of folds:", len(folds))


# Define the baseline and three machine learning models.
models = {

    # Baseline predicts the most common training category.
    "Majority Baseline": DummyClassifier(
        strategy="most_frequent"
    ),

    # Logistic Regression estimates the attack class.
    "Logistic Regression": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=RANDOM_SEED
        )
    ),

    # Support Vector Machine separates traffic categories.
    "Support Vector Machine": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        SVC(
            kernel="rbf",
            class_weight="balanced"
        )
    ),

    # Random Forest combines multiple decision trees.
    "Random Forest": make_pipeline(
        SimpleImputer(strategy="median"),
        RandomForestClassifier(
            n_estimators=150,
            min_samples_leaf=2,
            class_weight="balanced_subsample",
            random_state=RANDOM_SEED,
            n_jobs=-1
        )
    )
}


# Create a list to store model predictions.
prediction_parts = []


# Evaluate the models using the same device partitions.
for fold_number, (train_idx, test_idx) in enumerate(
    folds,
    start=1
):

    print(f"\nStarting validation Fold {fold_number}")

    # Select training records.
    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]

    # Select testing records.
    X_test = X.iloc[test_idx]

    # Train and test each model.
    for model_name, model_template in models.items():

        print("Training:", model_name)

        # Create a new independent model for this fold.
        model = clone(model_template)

        # Fit the model using only training data.
        model.fit(X_train, y_train)

        # Predict labels for devices excluded from training.
        predictions = model.predict(X_test)

        # Preserve the actual labels and device information.
        results = sample_df.iloc[test_idx][
            [
                "device",
                "family",
                "source_file",
                "label"
            ]
        ].copy()

        # Record the validation fold.
        results["fold"] = fold_number

        # Record which model produced the prediction.
        results["model"] = model_name

        # Store the predicted traffic category.
        results["prediction"] = predictions

        # Add these predictions to the results collection.
        prediction_parts.append(results)

        print("Completed:", model_name)


# Combine predictions from all validation folds.
predictions_df = pd.concat(
    prediction_parts,
    ignore_index=True
)


# Display the number of predictions collected.
print("\nMachine learning comparison completed.")

print("Total prediction records:", len(predictions_df))

print("\nPredictions by model:")
print(predictions_df["model"].value_counts())

print("\nDevice held out validation completed successfully.")

# Step 7 will calculate the performance metrics.
```

Step 7: Evaluate Detection Performance

Each model is evaluated using accuracy, balanced accuracy, precision, recall, F1 score, and confusion matrix results.

False positives represent normal traffic incorrectly identified as malicious. False negatives represent malicious traffic missed by the model.

Performance is calculated from the held-out device predictions. The majority baseline provides a reference for comparing trained model results.

```python
# Step 7: Calculate detection performance

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Define one consistent evaluation function.
def evaluate_predictions(data):

    actual = data["label"]
    predicted = data["prediction"]

    # Read the confusion matrix in a fixed class order.
    tn, fp, fn, tp = confusion_matrix(
        actual,
        predicted,
        labels=[0, 1]
    ).ravel()

    # Calculate the false-positive rate.
    false_positive_rate = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else float("nan")
    )

    return {
        "accuracy": accuracy_score(actual, predicted),
        "balanced_accuracy": balanced_accuracy_score(
            actual, predicted
        ),
        "precision": precision_score(
            actual, predicted, zero_division=0
        ),
        "recall": recall_score(
            actual, predicted, zero_division=0
        ),
        "f1_score": f1_score(
            actual, predicted, zero_division=0
        ),
        "true_positives": int(tp),
        "true_negatives": int(tn),
        "false_positives": int(fp),
        "false_negatives": int(fn),
        "false_positive_rate": false_positive_rate
    }


# Calculate results across all held-out predictions.
pooled_rows = []

for model_name, group in predictions_df.groupby("model"):

    scores = evaluate_predictions(group)
    scores["model"] = model_name

    pooled_rows.append(scores)

pooled_metrics = pd.DataFrame(pooled_rows)

# Order the comparison by balanced accuracy.
pooled_metrics = pooled_metrics.sort_values(
    "balanced_accuracy",
    ascending=False
)

print("\nHeld-out device model comparison:")
print(
    pooled_metrics.round(4).to_string(index=False)
)

# Calculate separate results for each validation fold.
fold_rows = []

for (fold_number, model_name), group in (
    predictions_df.groupby(["fold", "model"])
):

    scores = evaluate_predictions(group)

    scores["fold"] = fold_number
    scores["model"] = model_name

    fold_rows.append(scores)

fold_metrics = pd.DataFrame(fold_rows)

print("\nPerformance by validation fold:")
print(
    fold_metrics[
        ["fold", "model", "balanced_accuracy",
         "precision", "recall", "f1_score"]
    ].round(4).to_string(index=False)
)
```

Step 8: Examine Errors and Generalization

Overall model performance may hide differences between devices or attack families.

Detection results are examined separately for each device and botnet family. This identifies whether models produce false alerts on particular devices or miss malicious traffic associated with specific attack families.

Variation across devices is important when evaluating whether a model could generalize beyond the conditions represented in its training data.

```python
# Step 8: Examine model errors across devices

# Calculate performance for each device and model.
device_rows = []

for (model_name, device), group in (
    predictions_df.groupby(["model", "device"])
):

    scores = evaluate_predictions(group)

    scores["model"] = model_name
    scores["device"] = device

    device_rows.append(scores)

device_metrics = pd.DataFrame(device_rows)

print("\nDetection results by device:")

print(
    device_metrics[
        ["model", "device", "balanced_accuracy",
         "recall", "false_positive_rate",
         "false_positives", "false_negatives"]
    ].round(4).to_string(index=False)
)

# Examine only the records labeled as malicious.
attack_predictions = predictions_df[
    predictions_df["label"] == 1
]

# Calculate recall separately for each botnet family.
family_metrics = (
    attack_predictions
    .groupby(["model", "family"])
    .agg(
        attack_recall=("prediction", "mean"),
        attack_records=("prediction", "size")
    )
    .reset_index()
)

print("\nAttack detection by botnet family:")
print(
    family_metrics.round(4).to_string(index=False)
)

# Compare balanced accuracy across devices.
device_comparison = device_metrics.pivot(
    index="device",
    columns="model",
    values="balanced_accuracy"
)

print("\nDevice generalization comparison:")
print(device_comparison.round(4))

print("\nError and generalization review completed.")
```

Results and Interpretation

The N BaIoT experiment evaluated 7,200 network traffic records from nine IoT devices, including 3,600 benign and 3,600 malicious observations. Three-fold device grouped validation was used to assess detection performance on unfamiliar devices.

Random Forest achieved the highest accuracy at 99.83%, followed by Logistic Regression at 99.60% and Support Vector Machine at 99.21%. The majority class baseline achieved 50.00% accuracy.

Random Forest correctly identified 3,599 of 3,600 malicious records, with 11 false positives and one false negative. Although overall detection performance was high, errors occurred on two security cameras, showing that results varied across devices.

H3 Research Contribution

The experiment demonstrates that machine learning can identify recorded IoT botnet activity under the evaluated device held out conditions. Random Forest achieved the strongest overall performance among the three trained models.

These findings provide supplemental evidence about the network detection component of H3. However, identifying malicious traffic does not establish that cybersecurity analytics reduced baggage tracking disruptions.

Limitations

The N BaIoT dataset contains commercial IoT network observations rather than aviation baggage tracking records. The experiment used a balanced sample and evaluated known botnet families.

Device grouped validation reduced overlap between training and testing devices but did not establish performance against completely new attack families or independent network environments.

No baggage scan delivery, operational recovery, or adaptive mitigation was measured in this experiment.

Final Assessment

Notebook 13 provides supplemental evidence supporting the feasibility of machine learning network attack detection. The results do not independently confirm H3 because operational denial of service mitigation was not evaluated.

The existing H3 operational simulations remain the primary evidence for evaluating modeled baggage scan delivery, delays, and recovery under protective response conditions.
